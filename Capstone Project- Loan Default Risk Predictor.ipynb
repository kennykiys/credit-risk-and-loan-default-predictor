## Replacing Missing Numerical Values
### Several numerical variables contained missing values and exhibited outliers. Median imputation was used because the median is robust to extreme values and provides a more representative measure of central tendency for skewed financial data. Using the mean could have produced biased imputations due to the presence of outliers.


```python
import numpy as np

# Select numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Fill missing values with the median
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

## Replacing missing categorical Values
#### Several categorical variables contained missing values. Mode imputation was used because the mode represents the most frequently occurring category and preserves the existing distribution of the data. Using the mode provides a simple and reasonable estimate for missing categorical values without introducing invalid or artificial categories.

```python
# Select categorical columns
cat_cols = df.select_dtypes(include='object').columns

# Replace missing values with the most frequent category
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

## **Hyperparameter Tuning**

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score


log_reg_tuned = LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)

log_param_grid = {
    "C": [0.01, 0.1, 1, 10],
    "solver": ["liblinear", "lbfgs"]
}


log_grid = GridSearchCV(
    estimator=log_reg_tuned,
    param_grid=log_param_grid,
    scoring="recall",
    cv=5,
    n_jobs=-1,
    verbose=2
)


log_grid.fit(X_train_scaled, y_train)


best_log = log_grid.best_estimator_

y_pred_log_tuned = best_log.predict(X_test_scaled)


print("Best Parameters:", log_grid.best_params_)
print("Best CV Recall:", log_grid.best_score_)

print("\nAccuracy:")
print(accuracy_score(y_test, y_pred_log_tuned))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_log_tuned))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_log_tuned))
```

```python
from sklearn.tree import DecisionTreeClassifier


dt_tuned = DecisionTreeClassifier(
    class_weight="balanced",
    random_state=42
)


dt_param_grid = {
    "criterion": ["gini", "entropy"],
    "max_depth": [5, 10, 20, None],
    "min_samples_split": [2, 10, 20],
    "min_samples_leaf": [1, 5, 10]
}


dt_grid = GridSearchCV(
    estimator=dt_tuned,
    param_grid=dt_param_grid,
    scoring="recall",
    cv=5,
    n_jobs=-1,
    verbose=2
)


dt_grid.fit(X_train, y_train)


best_dt = dt_grid.best_estimator_

y_pred_dt_tuned = best_dt.predict(X_test)


print("Best Parameters:", dt_grid.best_params_)
print("Best CV Recall:", dt_grid.best_score_)

print("\nAccuracy:")
print(accuracy_score(y_test, y_pred_dt_tuned))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt_tuned))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_dt_tuned))
```

```python
from sklearn.ensemble import RandomForestClassifier


rf_tuned = RandomForestClassifier(
    class_weight="balanced",
    random_state=42
)


rf_param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [10, 20, None],
    "min_samples_split": [2, 10],
    "min_samples_leaf": [1, 5]
}


rf_grid = GridSearchCV(
    estimator=rf_tuned,
    param_grid=rf_param_grid,
    scoring="recall",
    cv=5,
    n_jobs=-1,
    verbose=2
)


rf_grid.fit(X_train, y_train)


best_rf = rf_grid.best_estimator_

y_pred_rf_tuned = best_rf.predict(X_test)


print("Best Parameters:", rf_grid.best_params_)
print("Best CV Recall:", rf_grid.best_score_)

print("\nAccuracy:")
print(accuracy_score(y_test, y_pred_rf_tuned))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf_tuned))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf_tuned))
```

#### Interpretation of Tuned Model Results

### Final Model Selection

After comparing the tuned models, Logistic Regression was selected as the final model for the loan default prediction system.
While Random Forest achieved the highest overall accuracy (**72.7%**) and F1-score (**0.27**), this project prioritizes **recall** over raw accuracy — in a lending context, failing to identify an applicant who will default (a false negative) is more costly than flagging a safe applicant for extra review (a false positive). Accuracy alone is also a misleading metric here, given the significant class imbalance in the dataset.

On that basis, Logistic Regression stood out:

* Strong recall (**68%**)
* Higher accuracy (**68.5%**) than Decision Tree
* Better balance between detecting defaults and staying interpretable for real-world lending decisions

Therefore, the tuned Logistic Regression was selected because it provides the most practical balance for a real-world loan default prediction system.

```python
# Import libraries
import tensorflow as tf
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.utils import class_weight
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix


# Calculate class weights to handle imbalance
weights = class_weight.compute_class_weight(
    class_weight="balanced",
    classes=np.unique(y_train),
    y=y_train
)

class_weights = {
    0: weights[0],
    1: weights[1]
}

print("Class weights:", class_weights)


# Build ANN model
ann = Sequential([
    Dense(64, activation="relu", input_shape=(X_train_scaled.shape[1],)),
    Dropout(0.3),

    Dense(32, activation="relu"),

    Dense(1, activation="sigmoid")
])


# Compile model
ann.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


# Stop training if validation performance stops improving
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)


# Train ANN
history = ann.fit(
    X_train_scaled,
    y_train,
    validation_split=0.2,
    epochs=30,
    batch_size=32,
    callbacks=[early_stop],
    class_weight=class_weights,
    verbose=1
)


# Predict probabilities
y_prob = ann.predict(X_test_scaled)


# Convert probabilities into classes
y_pred_ann = (y_prob >= 0.5).astype(int)


# Evaluate model
print("\nAccuracy:")
print(accuracy_score(y_test, y_pred_ann))


print("\nClassification Report:")
print(classification_report(y_test, y_pred_ann))


print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_ann))
```
